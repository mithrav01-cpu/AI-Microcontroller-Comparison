```python
print("AI-Based Microcontroller Platform Comparison")
print("=" * 50)

platforms = ["Arduino Uno", "ESP32", "Raspberry Pi Pico"]

for platform in platforms:
    print(platform)
```

```python
import pandas as pd

data = {
    "Platform": ["Arduino Uno", "ESP32", "Raspberry Pi Pico"],
    "RAM_KB": [2, 520, 264],
    "Flash_KB": [32, 4096, 2048],
    "Cost_INR": [500, 400, 400]
}

df = pd.DataFrame(data)

print("MICROCONTROLLER COMPARISON DATASET")
display(df)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.bar(df["Platform"], df["RAM_KB"])
plt.title("RAM Comparison")
plt.xlabel("Microcontroller")
plt.ylabel("RAM (KB)")
plt.show()

plt.figure(figsize=(8, 5))
plt.bar(df["Platform"], df["Flash_KB"])
plt.title("Flash Memory Comparison")
plt.xlabel("Microcontroller")
plt.ylabel("Flash Memory (KB)")
plt.show()

plt.figure(figsize=(8, 5))
plt.bar(df["Platform"], df["Cost_INR"])
plt.title("Cost Comparison")
plt.xlabel("Microcontroller")
plt.ylabel("Cost (INR)")
plt.show()
```

```python
df.to_csv("microcontroller_dataset.csv", index=False)

print("Dataset saved successfully!")

from google.colab import files
files.download("microcontroller_dataset.csv")
```

```python
from sklearn.tree import DecisionTreeClassifier

X = [
    [2, 32, 500],
    [520, 4096, 400],
    [264, 2048, 400]
]

y = [0, 1, 2]

model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

print("Machine Learning Model Trained Successfully!")
```

```python
platform_names = {
    0: "Arduino Uno",
    1: "ESP32",
    2: "Raspberry Pi Pico"
}

ram = 520
flash = 4096
cost = 400

prediction = model.predict([[ram, flash, cost]])

print("RAM:", ram, "KB")
print("Flash Memory:", flash, "KB")
print("Budget: Rs.", cost)
print("Recommended Platform:", platform_names[prediction[0]])
```
